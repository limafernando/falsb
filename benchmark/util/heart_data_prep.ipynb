```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from imblearn.over_sampling import SMOTE
```

```python
data = pd.read_csv('../data/heart/heart.csv')
```

```python
data.head()
```

```python
data = data.dropna()
```

```python
data["sex"].value_counts()
```

```python
data['sex'] = data['sex'].map({1: 0, 0: 1}) # inverse
```

```python
data["sex"].value_counts()
```

```python
len(data[(data['age'] >= 45) & (data['age'] <= 65)])
```

```python
data['age'] = ((data['age'] < 45) | (data['age'] > 65)).astype(int) # not middle aged
```

```python
len(data[data["age"] == 0].index)
```

```python
data.head()
```

```python
scaler = MinMaxScaler()
scaler.fit(data[["trestbps"]])
data["trestbps"] = scaler.transform(data[["trestbps"]])
```

```python
scaler = MinMaxScaler()
scaler.fit(data[["chol"]])
data["chol"] = scaler.transform(data[["chol"]])
```

```python
scaler = MinMaxScaler()
scaler.fit(data[["thalach"]])
data["thalach"] = scaler.transform(data[["thalach"]])
```

```python
scaler = MinMaxScaler()
scaler.fit(data[["oldpeak"]])
data["oldpeak"] = scaler.transform(data[["oldpeak"]])
```

```python
data.head()
```

```python
categorical_attr = ['cp', 'restecg', 'slope', 'ca', 'thal']
```

```python
for attr in categorical_attr:
    column_idx = data.columns.get_loc(attr)
    data = pd.concat([data, pd.get_dummies(data[attr], prefix=attr, dtype=int)], axis=1)
```

```python
data.head()
```

```python
columns_order = [
    "age",
    "sex",
    "fbs",
    "restecg",
    "exang",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak",
    "cp_0",
    "cp_1",
    "cp_2",
    "cp_3",
    "slope_0",
    "slope_1",
    "slope_2",
    "ca_0",
    "ca_1",
    "ca_2",
    "ca_3",
    "ca_4",
    "thal_0",
    "thal_1",
    "thal_2",
    "thal_3",
    "target",
]
```

```python
len(columns_order)
```

```python
data = data[columns_order]
data.head()
```

```python
print(len(data.index))
```

```python
data["target"].value_counts()
```

```python
data.to_csv('../data/heart/post_prep/heart.csv')
```
