```python
import pandas as pd
from sklearn.preprocessing import Binarizer
from imblearn.over_sampling import SMOTE
```

```python
data = pd.read_csv('../data/stroke/stroke.csv')
```

```python
data.head()
```

```python
data = data.dropna()
```

```python
data = data.drop("id", axis="columns")
```

```python
data = data[data['gender'] != 'Other']
```

```python
data["gender"].value_counts()
```

```python
data['gender'] = pd.get_dummies(data['gender'], dtype=int)['Female']
```

```python
data["gender"].value_counts()
```

```python
len(data[data["age"] >= 60].index)
```

```python
binarizer = Binarizer(threshold=59)
data['age'] = binarizer.fit_transform(data[['age']])
```

```python
len(data[data["age"] == 1].index)
```

```python
data["ever_married"].value_counts()
```

```python
data['ever_married'] = pd.get_dummies(data['ever_married'], dtype=int)['Yes']
```

```python
data["ever_married"].value_counts()
```

```python
data["Residence_type"].value_counts()
```

```python
data['Residence_type'] = pd.get_dummies(data['Residence_type'], dtype=int)['Urban']
```

```python
data["Residence_type"].value_counts()
```

```python
categorical_attr = ['work_type', 'smoking_status']
```

```python
for attr in categorical_attr:
    column_idx = data.columns.get_loc(attr)
    data = pd.concat([data, pd.get_dummies(data[attr], prefix=attr, dtype=int)], axis=1)
```

```python
data = data.drop(labels=categorical_attr, axis=1)
```

```python
data.head()
```

```python
columns_order = [
    "gender",
    "age",
    "hypertension",
    "heart_disease",
    "ever_married",
    "Residence_type",
    "work_type_Govt_job",
    "work_type_Never_worked",
    "work_type_Private",
    "work_type_Self-employed",
    "work_type_children",
    "smoking_status_Unknown",
    "smoking_status_formerly smoked",
    "smoking_status_never smoked",
    "smoking_status_smokes",
    "avg_glucose_level",
    "bmi",
    "stroke"
]
```

```python
len(columns_order)
```

```python
data = data[columns_order]
data.head()
```

```python
print(len(data.index))
```

```python
data.to_csv('../data/stroke/post_prep/stroke.csv')
```

```python
sm = SMOTE(random_state=42)
x, y = data.iloc[:, :-1], data.iloc[:, -1]
X_res, y_res = sm.fit_resample(x, y)
```

```python
y_res
```

```python
balanced_data = X_res
balanced_data["stroke"] = y_res
```

```python
balanced_data.head()
```

```python
balanced_data.to_csv('../data/stroke/post_prep/balanced_stroke.csv')
```
