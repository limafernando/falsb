```python
df = pd.read_csv("../benchmark/data/stroke/stroke.csv", delimiter=",")
```

```python
stroke = df[df['stroke'] == 1]
man_stroke = len(stroke[stroke['gender'] == "Male"].index)
woman_stroke = len(stroke[stroke['gender'] == "Female"].index)

non_stroke = df[df['stroke'] == 0]
man_non_stroke = len(non_stroke[non_stroke['gender'] == "Male"].index)
woman_non_stroke = len(non_stroke[non_stroke['gender'] == "Female"].index)


res_df = pd.DataFrame(
    {
        'non stroke': [man_non_stroke, woman_non_stroke],
        'stroke': [man_stroke, woman_stroke],
    },
    index=['male', 'female']
)

res_df.plot(kind='bar', stacked=True)#, color=['red', 'pink'])
plt.legend(loc="upper right")
plt.xticks(rotation=0)
plt.show()
```

```python
stroke = df[df['stroke'] == 1]
elder_stroke = len(stroke[stroke['age'] >= 60].index)
younger_stroke = len(stroke[stroke['age'] < 60].index)

non_stroke = df[df['stroke'] == 0]
elder_non_stroke = len(non_stroke[non_stroke['age'] >= 60].index)
younger_non_stroke = len(non_stroke[non_stroke['age'] < 60].index)


res_df = pd.DataFrame(
    {
        'non stroke': [elder_non_stroke, younger_non_stroke],
        'stroke': [elder_stroke, younger_stroke],
    },
    index=['elder', 'younger']
)

res_df.plot(kind='bar', stacked=True)#, color=['red', 'pink'])
plt.legend(loc="upper right")
plt.xticks(rotation=0)
plt.show()
```
